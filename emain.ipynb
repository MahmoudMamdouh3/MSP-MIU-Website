```python
import pandas as pd

def clean_duplicates(df):
    # تنظيف أسماء الأعمدة من مسافات
    df.columns = df.columns.str.strip()

    print("أسماء الأعمدة قبل التنظيف:")
    print(df.columns.tolist())

    # حذف الأعمدة المكررة بالاسم
    df = df.loc[:, ~df.columns.duplicated()]

    # حذف الصفوف المكررة تماماً
    before = len(df)
    df = df.drop_duplicates(keep='first')
    print(f"تم حذف {before - len(df)} صفوف مكررة (محتوى مكرر بالكامل)")

    # حذف المكرر بناءً على الإيميل
    if 'MIU Email' in df.columns:
        before = len(df)
        df = df.drop_duplicates(subset=['MIU Email'], keep='first')
        print(f"تم حذف {before - len(df)} صفوف مكررة بناءً على البريد الإلكتروني")

    # حذف المكرر بناءً على الـ student_id
    if 'MIU University ID' in df.columns:
        before = len(df)
        df = df.drop_duplicates(subset=['MIU University ID'], keep='first')
        print(f"تم حذف {before - len(df)} صفوف مكررة بناءً على MIU University ID")

    return df


if __name__ == "__main__":
    merged_path = "data/merged_with_status_from_notebook.xlsx"
    df = pd.read_excel(merged_path)
    df_cleaned = clean_duplicates(df)
    df_cleaned.to_excel("data/merged_cleaned.xlsx", index=False)
    print("✅ تم تنظيف البيانات وحفظ الملف بدون أعمدة إضافية.")
```
